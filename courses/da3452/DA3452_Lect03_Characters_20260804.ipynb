```python
cafe_1 = "caf\u00E9"      # precomposed é
cafe_2 = "cafe\u0301"     # e + combining acute accent

print(cafe_1)
print(cafe_2)
```

They look the same. Let us check programmatically if they are indeed the same or not




```python
import unicodedata

a = "caf\u00E9"      # precomposed
b = "cafe\u0301"    # decomposed

print(a == b)
print(len(a), len(b))
```

Let us see what happens when we normalize this

```python
a = unicodedata.normalize("NFC", a)
b = unicodedata.normalize("NFC", b)

print(a == b)
print([f"U+{ord(c):04X}" for c in b])
```
